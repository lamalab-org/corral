```python
all_res = []
for name1, hex1 in PALLETT.items():
    res = []
    for name2, hex2 in PALLETT.items():
        if name1 != name2:
            rgb1 = sRGBColor.new_from_rgb_hex(hex1)
            lab1 = convert_color(rgb1, LabColor, target_illuminant='d65')

            rgb2 = sRGBColor.new_from_rgb_hex(hex2)
            lab2 = convert_color(rgb2, LabColor, target_illuminant='d65')

            cie = delta_e_cie2000(lab1, lab2)
            cmc = delta_e_cmc(lab1, lab2, pl=1)
            avg = round((cie + cmc)/2)
            res.append((name2, cie, cmc, avg))
    res.sort(key=lambda t: t[3])
    closest = res[0]
    all_res.append((name1, closest[0], closest[1], closest[2], closest[3]))

all_res.sort(key=lambda t: t[4])

for t in all_res:
    print(f"{t[0]}   {t[1]}     {t[2]:.2f}     {t[3]:.2f}     {t[4]:.2f}")
```
